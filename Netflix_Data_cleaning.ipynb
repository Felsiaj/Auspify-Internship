```python
import pandas as pd
```

```python
df = pd.read_csv("../Dataset/Dataset.csv")
df.head()
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
df.info()
```

```python
missing_values = df.isnull().sum()
print("Missing values in each column:")
print(missing_values)
```

```python
duplicate_count = df.duplicated().sum()
print("Number of duplicate rows:", duplicate_count)
```

```python
print("Type values:")
print(df["type"].unique())

print("\nRating values:")
print(df["rating"].unique())

print("\nSample country values:")
print(df["country"].head(10))
```

```python
# Check for leading or trailing spaces in text columns

text_columns = ["type", "title", "director", "country", "rating", "duration", "listed_in"]

for column in text_columns:
    extra_spaces = df[column].astype(str).str.strip().ne(df[column].astype(str)).sum()
    print(f"{column}: {extra_spaces} values with extra spaces")
```

```python
# Remove leading and trailing spaces from text columns

for column in text_columns:
    df[column] = df[column].astype(str).str.strip()

print("Extra spaces removed from text columns.")
```

```python
# Verify extra spaces after cleaning

for column in text_columns:
    extra_spaces = df[column].astype(str).str.strip().ne(df[column].astype(str)).sum()
    print(f"{column}: {extra_spaces} values with extra spaces")
```

```python
# Standardize the Type column

df["type"] = df["type"].str.strip().str.title()

print(df["type"].unique())
```

```python
# Correct the Type values

df["type"] = df["type"].replace({
    "Tv Show": "TV Show"
})

print(df["type"].unique())
```

```python
# Check unique rating values

print(df["rating"].unique())
```

### Rating Standardization

The rating column was checked for formatting inconsistencies. The values were already consistently formatted, so no changes were required.

```python
# Check unique country values

print(df["country"].unique())
```

### Country Standardization

The country column was checked for formatting inconsistencies. The values were consistently formatted, so no changes were required. The value "Not Given" was retained as it represents an unspecified country value in the dataset.

```python
# Check data types of all columns

print(df.dtypes)
```

```python
# Convert date_added to datetime format

df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")

print(df["date_added"].dtype)
```

```python
# Check for invalid dates

invalid_dates = df["date_added"].isna().sum()

print("Invalid date values:", invalid_dates)
```

```python
# Final check for missing values

print("Missing values in each column:")
print(df.isnull().sum())
```

```python
# Final check for duplicate rows

print("Number of duplicate rows:", df.duplicated().sum())
```

```python
# Export the cleaned dataset

output_path = "../Output/Cleaned_Netflix_Dataset.csv"

df.to_csv(output_path, index=False)

print("Cleaned dataset exported successfully!")
print("Saved to:", output_path)
```
